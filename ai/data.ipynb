```python
import pandas as pd
import ast

df = pd.read_csv('data_final_normalized_fixed_disaster.csv')

def encode_disaster_type(val_str):
    try:
        # Chuyển đổi chuỗi ký tự thành danh sách (list) và sau đó là tập hợp (set) để so sánh không phụ thuộc thứ tự
        # Ví dụ: "['lũ lụt', 'ngập nước']" -> {'lũ lụt', 'ngập nước'}
        val_list = ast.literal_eval(val_str)
        val_set = set(val_list)
    except:
        # Trường hợp không parse được thì giữ nguyên
        return val_str

    # Áp dụng các quy tắc encoding
    if val_set == {'không xác định'}:
        return 0
    elif val_set == {'lũ lụt'} or val_set == {'bão'} or val_set == {'sạt lở'} or val_set == {'hạn hán'} or val_set == {'ngập nước'}:
        return 1
    # 'lũ lụt' và 'ngập nước' (bất kể thứ tự)
    elif val_set == {'lũ lụt', 'ngập nước'} or val_set == {'ngập nước', 'lũ lụt'}:
        return 2
    # 'lũ lụt' và 'sạt lở' (bất kể thứ tự)
    elif val_set == {'lũ lụt', 'sạt lở'} or val_set == {'sạt lở', 'lũ lụt'}:
        return 3
    elif val_set == {'lũ lụt', 'bão'} or val_set == {'bão', 'lũ lụt'}:
        return 2
    elif val_set == {'ngập nước', 'không xác định'} or val_set == {'không xác định', 'ngập nước'}:
        return 1
    # ngập nước, sạt lở
    elif val_set == {'ngập nước', 'sạt lở'} or val_set == {'sạt lở', 'ngập nước'}:
        return 3
    else:
        # Giữ nguyên các giá trị khác không được quy định
        return val_str

# Áp dụng hàm encoding vào cột disaster_type
df['disaster_type'] = df['disaster_type'].apply(encode_disaster_type)

# Lưu kết quả ra file csv mới
df.to_csv('data_final_final.csv', index=False)

# Kiểm tra nhanh kết quả
print(df['disaster_type'].value_counts())
```
